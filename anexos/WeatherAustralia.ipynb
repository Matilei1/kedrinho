## Data cleaning

Limpiar los datos meteorológicos gestionando los valores faltantes y convirtiendo la columna "Fecha" en objetos de fecha y hora.


**Razonamiento**:
Imputar valores faltantes, convertir la columna 'Fecha' en objetos de fecha y hora, eliminar duplicados y manejar inconsistencias en otras columnas.



```python
import pandas as pd

# IMPUTACIÓN DE VALORES FALTANTES (MISSING VALUE IMPUTATION)
# Obtener los nombres de las columnas numéricas y categóricas
numerical_cols = df.select_dtypes(include=['number']).columns
categorical_cols = df.select_dtypes(include=['object']).columns


for col in numerical_cols:
    # Crear una columna binaria indicando si el valor original estaba ausente (1) o no (0)
    df[col + '_missing'] = df[col].isnull().astype(int)  # Create binary indicator
    # Rellenar los valores faltantes con la mediana de la columna
    df[col] = df[col].fillna(df[col].median())

for col in categorical_cols:
     # Crear indicador binario de valores faltantes
    df[col + '_missing'] = df[col].isnull().astype(int) # Create binary indicator
     # Rellenar los valores faltantes con la moda (valor más frecuente)
    df[col] = df[col].fillna(df[col].mode()[0])

# Date Conversion
# Convertir la columna 'Date' a tipo datetime
try:
    df['Date'] = pd.to_datetime(df['Date'])
except ValueError as e:
    print(f"Error converting 'Date' column: {e}")
    # Handle the error, e.g., by removing invalid rows or replacing with NaT
    # df = df[pd.to_datetime(df['Date'], errors='coerce').notnull()]
    # df['Date'] = pd.to_datetime(df['Date'], errors='coerce')

# Duplicate Removal
df = df.drop_duplicates()

# Inconsistency Handling
# Check for non-numeric values in numerical columns and correct them if necessary
#Verificar valores no numéricos en columnas numéricas y corregirlos si es necesario.
for col in numerical_cols:
    # Intentar convertir la columna a tipo numérico, forzando errores a NaN
    df[col] = pd.to_numeric(df[col], errors='coerce')
    # Rellenar los NaN generados con la mediana
    df[col] = df[col].fillna(df[col].median())

display(df.head())
```

## Resumen:

El proceso de limpieza de datos implicó varios pasos:

1.  **Missing Value Imputation:** Los valores numéricos faltantes se completaron con la mediana de cada columna, y los valores categóricos faltantes se completaron con la moda. Se crearon columnas de indicadores binarios para rastrear dónde faltaban valores originalmente.
2.  **Date Conversion:** La columna 'Date' se convirtió a objetos de fecha y hora.
3.  **Duplicate Removal:** Se eliminaron las filas duplicadas del conjunto de datos.
4.  **Inconsistency Handling:**  Los valores no numéricos en las columnas numéricas se convirtieron a NaN y luego se completaron con la mediana de la columna.


### Ánalisis de  los datos y hallazgos clave

*   **Missing Data:**  Se encontraron valores faltantes significativos en varias columnas (p. ej., 'Evaporation', 'Sunshine', 'Cloud9am'). Estos se imputaron utilizando la mediana para las características numéricas y la moda para las categóricas. Se crearon columnas de indicadores binarios para señalar los valores faltantes originales.

*   **Data Types:** La columna 'Date' era inicialmente un tipo de objeto y se convirtió correctamente a objetos de fecha y hora.
*   **Feature Engineering:** Las características "Month" y "Year" se diseñaron a partir de la columna "Date".
*   **Categorical Encoding:** Se aplicó la codificación one-hot a todas las características categóricas, incluyendo la gestión de los valores faltantes mediante la creación de una categoría independiente para cada una. Se eliminó la primera categoría de cada característica codificada one-hot para evitar la multicolinealidad.



```python
import pandas as pd

# Asegurarse de que 'Date' sea tipo datetime
df['Date'] = pd.to_datetime(df['Date'], errors='coerce')

# Manipulación de la fecha
df['Month'] = df['Date'].dt.month
df['Year'] = df['Date'].dt.year
df['Day'] = df['Date'].dt.day

# Mostrar 'Date', 'Year', 'Month' y 'Day' antes de codificar
display(df[['Date', 'Year', 'Month', 'Day']].head(10))

# FILTRAR LOCALIDADES PERMITIDAS
localidades_permitidas = [
  'Adelaide', 'Albury', 'Ballarat', 'Bendigo', 'Brisbane', 'Cairns', 'Canberra',
  'Darwin', 'GoldCoast', 'Launceston', 'Mildura', 'Moree', 'MountGambier',
  'Newcastle', 'Perth', 'Sale', 'Sydney', 'WaggaWagga', 'Wollongong'
]
df = df[df['Location'].isin(localidades_permitidas)]

# 🔍 Mostrar resultados después del filtrado
print('Tabla original con localidades permitidas')
print(f" El DataFrame tiene {df.shape[0]:,} filas y {df.shape[1]} columnas.")
display(df.head(10))



# ONE-HOT ENCODING para columnas categóricas (excepto Location)
categorical_cols = df.select_dtypes(include=['object']).columns.drop('Location')
df_encoded = pd.get_dummies(df, columns=categorical_cols, dummy_na=True, drop_first=True)



# ✅ Verificación del DataFrame codificado
print('Nuevo DataFrame con Locaciones filtradas y codificación aplicada')
display(df_encoded.head())
print(f" El nuevo DataFrame tiene {df_encoded.shape[0]:,} filas y {df_encoded.shape[1]} columnas.")

print("Cantidad total de filas (registros):", len(df_encoded))

print(sorted(df['Location'].unique()))
```

Se generaron nuevas columnas porque, en variables como **RainToday**, que originalmente eran de tipo object (categóricas), se splicó *one-hot encoding* es decir la categoría se convirtió en una columna binaria (0 o 1).
Sin embargo, incluir todas las categorías (No, Yes y NaN(faltante)) había redundandacia de datos por lo que se utilizo drop_first=True para eliminar la primera categoría alfabética (por defecto) y así la categoría puede ser representada sin tener redundancia y sin perder datos.

Por lo tanto, si tanto RainToday_Yes como RainToday_nan son False (es decir, ambos valen 0), significa que el valor original fue 'No', es decir, ese día no llovió.
